## Parte 2: Los Tres Tipos de Errores (Aprender a no temerle al error)

### 2.1. ¿Qué vamos a realizar?
En programación, equivocarse es completamente normal y es parte del aprendizaje diario. Existen **3 tipos de errores** fundamentales que todo futuro ingeniero ambiental debe saber reconocer:
1. **Errores de Sintaxis (*SyntaxError*):** Rompen las reglas ortográficas del lenguaje (olvidar un paréntesis, unas comillas, etc.).
2. **Errores de Ejecución (*RuntimeError*):** El código está bien escrito sintácticamente, pero le pide a la computadora una operación imposible durante la marcha (ejemplo: dividir para cero).
3. **Errores Lógicos:** El programa corre sin quejarse ni arrojar ninguna alerta roja, pero el resultado numérico es **incorrecto** porque la fórmula matemática fue mal planteada.

Veamos un ejemplo de cálculo correcto vs. un error lógico en el cálculo de un promedio de precipitación.


```python
# Cálculo del promedio de precipitación de 3 días en Loja:
# Datos en milímetros (mm):
dia1 = 12.0
dia2 = 18.0
dia3 = 15.0

# FORMA INCORRECTA (Error lógico por falta de paréntesis):
# En matemáticas, la división se evalúa antes que la suma:
promedio_incorrecto = dia1 + dia2 + dia3 / 3

# FORMA CORRECTA:
promedio_correcto = (dia1 + dia2 + dia3) / 3

print("Resultado con error lógico (fórmula mal agrupada):", promedio_incorrecto, "mm")
print("Resultado correcto:", promedio_correcto, "mm")
```

### 2.2. Explicación detallada:
* En `dia1 + dia2 + dia3 / 3`, la computadora divide primero `15.0 / 3 = 5.0` y luego suma `12.0 + 18.0 + 5.0 = 35.0 mm`. La máquina no sabe que querías un promedio; ella solo obedece la jerarquía matemática.
* En `(dia1 + dia2 + dia3) / 3`, los paréntesis fuerzan a que primero se sumen los tres valores (`45.0 mm`) y finalmente se divida entre 3, obteniendo el valor real de `15.0 mm`.
* **Lección clave:** Los errores lógicos son los más peligrosos en la ingeniería ambiental porque el programa no se detiene, pero entrega datos erróneos a los informes técnicos.

## Parte 3: Variables, Identificadores y Constantes

### 3.1. ¿Qué vamos a realizar?
Una **variable** es un espacio con nombre en la memoria del computador que almacena un dato que puede cambiar durante el tiempo (como la temperatura a lo largo del día).

* **Reglas para nombrar variables:**
  * Deben iniciar con una letra o guion bajo, nunca con un número (`2estacion` es inválido; `estacion_2` es válido).
  * No pueden contener espacios ni guiones medios (`altitud msnm` es error; usa `altitud_msnm`).
  * Python distingue entre mayúsculas y minúsculas: `Temperatura`, `temperatura` y `TEMPERATURA` son tres variables completamente distintas.

## Parte 4: Los Cuatro Tipos de Datos Primitivos

### 4.1. ¿Qué vamos a realizar?
En Python no necesitas declarar explícitamente el tipo de dato; el lenguaje lo infiere automáticamente (**tipado dinámico**). Sin embargo, como programador debes conocer con exactitud qué tipo de dato estás manipulando.

Los 4 tipos fundamentales son:
1. **`int` (Entero):** Números sin decimales (conteos de árboles, número de incendios, días del año).
2. **`float` (Flotante o Real):** Números con punto decimal (temperatura, precipitación, pH del suelo, coordenadas GPS).
3. **`str` (Cadena de texto / *String*):** Nombres, descripciones, categorías entre comillas.
4. **`bool` (Booleano):** Solo admite dos valores lógicos: `True` (Verdadero) o `False` (Falso).

Utilizaremos la función `type()` para pedirle a Python que nos revele el tipo interno de cada dato.


```python
# Inspección de tipos de datos en ciencias ambientales:
parroquia = "Malacatos"                # Cadena de texto (str)
numero_focos_calor = 7                 # Número entero (int)
velocidad_viento_kmh = 24.85           # Número decimal / flotante (float)
alerta_incendio_activada = True        # Valor booleano (bool)

# Consultamos el tipo de cada variable con type():
print("parroquia:", parroquia, "--> Tipo:", type(parroquia))
print("numero_focos_calor:", numero_focos_calor, "--> Tipo:", type(numero_focos_calor))
print("velocidad_viento_kmh:", velocidad_viento_kmh, "--> Tipo:", type(velocidad_viento_kmh))
print("alerta_incendio_activada:", alerta_incendio_activada, "--> Tipo:", type(alerta_incendio_activada))
```

### 4.2. Explicación detallada:
* `type(variable)` devuelve la clase a la que pertenece la información: `<class 'str'>`, `<class 'int'>`, `<class 'float'>`, `<class 'bool'>`.
* En Python los decimales se escriben obligatoriamente con **punto (`.`)** y nunca con coma (escribir `24,85` crearía una tupla de dos números en lugar de un flotante).
* Los booleanos `True` y `False` deben escribirse con la primera letra en **mayúscula**.

## Parte 5: Operadores Aritméticos y Fórmulas Ambientales

### 5.1. ¿Qué vamos a realizar?
Vamos a realizar cálculos matemáticos en Python utilizando todos sus operadores aritméticos:
* `+` : Suma
* `-` : Resta
* `*` : Multiplicación
* `/` : División real (siempre devuelve un `float`)
* `//`: División entera (descarta los decimales)
* `%` : Módulo o residuo de una división entera
* `**`: Potencia (exponente)

Aplicaremos estos operadores a un caso real de ingeniería: **la conversión de temperatura de grados Celsius a Fahrenheit** ($F = C \times \frac{9}{5} + 32$) y la distribución de horas de monitoreo.


```python
# =====================================================================
# Operaciones aritméticas y fórmulas ambientales
# =====================================================================

temp_celsius = 25.0

# 1. Conversión de temperatura a Fahrenheit:
temp_fahrenheit = (temp_celsius * 9 / 5) + 32

# 2. Operaciones con horas de monitoreo de campo:
total_horas_monitoreo = 58
duracion_turno = 8

turnos_completos = total_horas_monitoreo // duracion_turno  # División entera
horas_sobrantes = total_horas_monitoreo % duracion_turno    # Módulo (residuo)

# 3. Potenciación: cálculo del área de influencia de un sensor circular (Area = pi * r^2)
PI = 3.141592
radio_cobertura_km = 4.5
area_cobertura_km2 = PI * (radio_cobertura_km ** 2)

print("Temperatura en Celsius:", temp_celsius, "°C")
print("Temperatura equivalente en Fahrenheit:", temp_fahrenheit, "°F")
print("--------------------------------------------------")
print("Total de horas de monitoreo:", total_horas_monitoreo)
print("Turnos completos de 8 horas:", turnos_completos)
print("Horas restantes para el siguiente turno:", horas_sobrantes)
print("--------------------------------------------------")
print("Área de monitoreo del sensor:", area_cobertura_km2, "km²")
```

### 5.2. Explicación detallada:
* `temp_celsius * 9 / 5 + 32`: Primero se efectúa la multiplicación y división, y al final la suma con 32.
* `//` (División entera): `58 // 8` da `7`, porque el 8 cabe 7 veces enteras en 58.
* `%` (Módulo): `58 % 8` da `2`, que es el residuo sobrante ($7 \times 8 = 56$, sobran 2). Este operador es muy útil para saber si un número es par o impar, o para ciclos de tiempo.
* `**` (Potencia): En Python la potencia se escribe con doble asterisco `**` (no se usa el símbolo `^`, el cual en Python tiene otro significado lógico llamado XOR a nivel de bits).

## Parte 6: Operadores Relacionales y Lógicos (Condiciones de Alerta)

### 6.1. ¿Qué vamos a realizar?
Para que un programa tome decisiones o clasifique riesgos ambientales, primero debe comparar valores numéricos y evaluar condiciones lógicas.

* **Operadores Relacionales:** devuelven `True` o `False`:
  * `==` : ¿Es exactamente igual? (Cuidado: doble igual, no confundir con `=` de asignación)
  * `!=` : ¿Es diferente de?
  * `>` y `>=`: Mayor que / Mayor o igual que
  * `<` y `<=`: Menor que / Menor o igual que
* **Operadores Lógicos:** combinan múltiples condiciones:
  * `and`: Es verdadero solo si **ambas** condiciones se cumplen simultáneamente.
  * `or` : Es verdadero si **al menos una** de las condiciones se cumple.
  * `not`: Invierte el valor lógico (lo verdadero lo hace falso y viceversa).

Evaluaremos la conocida regla meteorológica de riesgo de incendios forestales (**Regla del 30-30-30**: Temperatura $> 30$ °C, Humedad relativa $< 30$ % y Viento $> 30$ km/h).


```python
# Datos meteorológicos captados por una estación en Loja:
temperatura = 31.5        # °C
humedad_relativa = 26.0   # %
velocidad_viento = 34.0   # km/h

# Evaluamos cada condición individualmente:
condicion_temp = temperatura > 30.0
condicion_humedad = humedad_relativa < 30.0
condicion_viento = velocidad_viento > 30.0

# Evaluamos la regla completa con el operador 'and' (deben cumplirse las tres):
alerta_critica_30_30_30 = condicion_temp and condicion_humedad and condicion_viento

# Evaluamos una alerta moderada con el operador 'or' (si al menos hace mucho calor o hay viento fuerte):
alerta_moderada = condicion_temp or condicion_viento

print("¿La temperatura supera los 30 °C?:", condicion_temp)
print("¿La humedad es menor al 30 %?:", condicion_humedad)
print("¿El viento supera los 30 km/h?:", condicion_viento)
print("==================================================")
print("¿ESTADO DE ALERTA CRÍTICA (30-30-30)?:", alerta_critica_30_30_30)
print("¿ESTADO DE ALERTA PREVENTIVA MODERADA?:", alerta_moderada)
```

### 6.2. Explicación detallada:
* Cada comparación `temperatura > 30.0` genera un valor booleano (`True`).
* `and` es muy estricto: requiere que la condición 1 **y** la condición 2 **y** la condición 3 sean verdaderas. Si una sola falla, el resultado final es `False`.
* `or` es flexible: con que una de las condiciones sea verdadera, el resultado final es `True`.
* Esto sienta las bases para las estructuras condicionales (`if`, `elif`, `else`) que se abordarán en la siguiente unidad.

## Parte 7: Entrada de Datos (`input()`), Conversión de Tipos y Cadenas F-String

### 7.1. ¿Qué vamos a realizar?
Hasta ahora hemos escrito los datos directamente en el código (*hardcoding*). Pero un programa útil debe ser capaz de solicitar datos al usuario o al técnico de campo en tiempo real.

Para esto usamos la función `input()`.  
> ⚠️ **REGLA DE ORO DE PYTHON:**  
> La función `input()` **SIEMPRE devuelve una cadena de texto (`str`)**, incluso si el usuario teclea números.  
> Si necesitas hacer cálculos matemáticos, debes convertir (*castear*) esa entrada a número usando `int()` o `float()`.

También aprenderemos a mostrar resultados elegantes con **f-strings** (`f"Texto con {variable:.2f}"`), controlando la cantidad de decimales en pantalla.


```python
# Demostración del comportamiento de input() y conversión de tipos:

# Supongamos que ingresamos los valores directamente para simular la entrada de teclado:
entrada_lluvia_texto = "14.8"   # Esto es lo que recibe input() internamente

# Intentar sumar directamente texto produce concatenación (pegar letras):
# "14.8" + "10" resultaría en "14.810" (¡Error común!)

# Conversión correcta al tipo float:
lluvia_mm = float(entrada_lluvia_texto)
lluvia_adicional = 5.2

lluvia_total = lluvia_mm + lluvia_adicional

# Formato moderno y elegante con f-strings:
nombre_sector = "Vilcabamba"

print(f"Sector evaluado: {nombre_sector}")
print(f"Precipitación registrada: {lluvia_mm:.2f} mm")
print(f"Precipitación adicional estimada: {lluvia_adicional:.2f} mm")
print(f"Precipitación acumulada final: {lluvia_total:.2f} mm")
```

### 7.2. Explicación detallada:
* `float("14.8")` transforma el texto `"14.8"` en el número real `14.8` apto para operar matemáticamente.
* Si el dato fuera una cantidad entera de personas o árboles, usaríamos `int("10")`.
* **Las f-strings (`f"..."`):** Al colocar la letra `f` antes de las comillas, puedes insertar variables directamente dentro del texto entre llaves `{variable}`.
* `{lluvia_total:.2f}` le indica a Python: *"muestra este número con exactamente 2 decimales"*. Esto es fundamental para reportes técnicos de ingeniería con cifras significativas.

## Parte 8: Proyecto Integrado: Entrada $\to$ Proceso $\to$ Salida

### 8.1. ¿Qué vamos a realizar?
Vamos a integrar todo lo aprendido en un algoritmo completo de ingeniería ambiental:  
**Cálculo del Balance Hídrico Simplificado y Monitoreo de Déficit de Humedad para una Microcuenca del Cantón Loja.**

* **Entradas:**
  1. Nombre de la microcuenca.
  2. Precipitación mensual observada ($P$, en mm).
  3. Evapotranspiración potencial estimada ($ETP$, en mm).
  4. Área de la microcuenca en hectáreas ($ha$).
* **Procesos:**
  1. Calcular el balance hídrico neto ($BH = P - ETP$).
  2. Determinar si existe déficit hídrico ($BH < 0$).
  3. Convertir el volumen neto a metros cúbicos ($1 \text{ mm en } 1 \text{ ha} = 10 \text{ m}^3$).
* **Salida:**
  1. Generar un informe técnico en consola con formato profesional para la toma de decisiones.


```python
# =====================================================================
# PROGRAMA: Balance Hídrico y Monitoreo de Déficit en Microcuencas
# Contexto: Cantón Loja - Gestión de Recursos Hídricos
# =====================================================================

# 1. ENTRADA DE DATOS (Variables del problema):
nombre_cuenca = "Microcuenca Malacatos - Sector El Carmen"
precipitacion_mm = 65.4       # Lluvia en el mes (mm)
evapotranspiracion_mm = 88.2  # Pérdida de agua por evaporación y plantas (mm)
area_hectareas = 450.0        # Área de captación (ha)

# 2. PROCESAMIENTO:
# Balance hídrico neto:
balance_hidrico_mm = precipitacion_mm - evapotranspiracion_mm

# Evaluación de condición de déficit:
existe_deficit = balance_hidrico_mm < 0.0

# Conversión a volumen en metros cúbicos (1 mm en 1 ha equivale a 10 m³ de agua):
# Volumen neto = Balance (mm) * Área (ha) * 10
volumen_neto_m3 = balance_hidrico_mm * area_hectareas * 10.0

# 3. SALIDA DE RESULTADOS (Reporte técnico):
print("=====================================================================")
print("          INFORME TÉCNICO DE BALANCE HÍDRICO MENSUAL                 ")
print("               Carrera de Ingeniería Ambiental - UNL                ")
print("=====================================================================")
print(f"Zona de estudio:             {nombre_cuenca}")
print(f"Área territorial:            {area_hectareas:,.1f} hectáreas")
print(f"Precipitación mensual (P):   {precipitacion_mm:.1f} mm")
print(f"Evapotranspiración (ETP):    {evapotranspiracion_mm:.1f} mm")
print("---------------------------------------------------------------------")
print(f"Balance Hídrico Neto:        {balance_hidrico_mm:.2f} mm")
print(f"¿Existe déficit de agua?:    {existe_deficit}")
print(f"Volumen hídrico resultante:  {volumen_neto_m3:,.2f} m³ de agua")
print("=====================================================================")
```

### 8.2. Explicación detallada del flujo completo:
1. **Fase de Entrada:** Definimos los parámetros físicos de la cuenca con nombres explícitos y unidades en comentarios (`_mm`, `_hectareas`).
2. **Fase de Proceso:**
   * Al restar `65.4 - 88.2` obtenemos un balance negativo de `-22.80 mm`.
   * La comparación `balance_hidrico_mm < 0.0` evalúa a `True`, señalando que se evapora más agua de la que llueve (déficit hídrico).
   * Multiplicamos el déficit por el área y por el factor $10$, calculando que faltan `-102,600.00 m³` de agua en la cuenca durante ese periodo.
3. **Fase de Salida:**
   * Empleamos `{area_hectareas:,.1f}` donde la coma `,` formatea los millares automáticamente y `.1f` fija un decimal.
